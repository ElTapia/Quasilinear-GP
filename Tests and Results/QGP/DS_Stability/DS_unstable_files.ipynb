```python
def init_gauss(x, A):
    output = w_dark_theo(x, 0) + A*np.exp(-x**2)
    return output

def sech(x):
    return 1/np.cosh(x)

def init_sech(x, A):
    output = w_dark_theo(x, 0) + A*sech(x)
    return output
```

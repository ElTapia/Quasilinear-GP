```python
import os
theo = load_theo_pickle(path_bright + "theo_bright_c_0.71_k_0.25_w_1_L_100.pck")  # rebuild vs. scipy version skew
# H1 error, computed from the saved final-state solutions (same Ns, dts/Nts as the L2 error above)
def bright_theo_full(x, t, c=np.sqrt(2)/2, w=1, x_0=0):
    return theo(x - c*t - x_0) * np.exp(-1j*(c**2/4*t + w*t))

errors_H1_br_moving = []
for N in Ns:
    folder_name = path_bright_time_mov + base + "{N}_order_{o}/".format(N=N, o=2)
    err_H1 = np.zeros(len(Nts))
    x = np.linspace(-L, L, N)
    h = np.abs(x[1] - x[0])
    for j, Nt in enumerate(Nts):
        Nt_int = int(Nt)
        sol_file = folder_name + "solution_Nt_{}/w_{:010d}.npy.gz".format(Nt_int, Nt_int)
        if not os.path.exists(sol_file):
            err_H1[j] = np.nan
            continue
        w_file = gzip.GzipFile(sol_file, "r")
        w_sim = np.load(w_file)
        diff = w_sim - bright_theo_full(x, tf)
        err_H1[j] = norm_H1_dx(diff, h)
    errors_H1_br_moving += [err_H1]
```

```python
# Regenerate the theoretical dark soliton with L_theo = 2L (matching what the bright
# case already does). The saved L=100 profile can't be evaluated far enough to cover
# x - c*t for a moving soliton without extrapolating past its own spline domain --
# that is the source of the spurious H1 growth seen at fine J. Cached to disk so this
# expensive regeneration only runs once.
L = 100
N_theo = 262_145
theo_name = path_D2 + "theo_c_0.57_k_-50_L_100_Ltheo200.pck"

if not os.path.exists(theo_name):
    theo_func, _ = soliton(c, k, soliton_type="dark", L=2*L, N=N_theo, as_func=True)
    with open(theo_name, "wb") as f:
        pickle.dump(theo_func, f)
else:
    theo_func = load_theo_pickle(theo_name)

def theo_dark(x, t):
    return theo_func(x - c*t)
```

```python
order = 2
tf = 5
L = 100
dts = np.array([2**(-j) for j in range(1, 14)])
Nts = tf/dts

Ns = np.array([500, 1_000, 2_000, 5_000, 10_000, 50_000, 100_000])

base = "errors_N_"

errors = []
errors_H1 = []
for N in Ns:
    folder_name = path_D2_time + base + "{N}_order_{o}/".format(N=N, o=2)
    err_L2 = np.zeros(len(Nts))
    err_H1 = np.zeros(len(Nts))
    x = np.linspace(-L, L, N)
    h = np.abs(x[1] - x[0])
    for j, Nt in enumerate(Nts):
        Nt_int = int(Nt)
        sol_file = folder_name + "solution_Nt_{}/w_{:010d}.npy.gz".format(Nt_int, Nt_int)
        if not os.path.exists(sol_file):
            err_L2[j] = np.nan
            err_H1[j] = np.nan
            continue
        w_file = gzip.GzipFile(sol_file, "r")
        w_sim = np.load(w_file)
        diff = w_sim - theo_dark(x, tf)
        err_L2[j] = np.sqrt(h) * np_norm(diff, 2)
        err_H1[j] = norm_H1_dx(diff, h)
    errors += [err_L2]
    errors_H1 += [err_H1]
```

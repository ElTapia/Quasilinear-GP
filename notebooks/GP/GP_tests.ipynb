# GP Dark solitons tests

```python
# Create solitons
L = 20

k = 0

pars = {"k":k}

c_1 = 0
c_2 = np.sqrt(2)/2

# Theretical solitons
w_black, x_black = soliton(c_1, k, soliton_type="dark", L=L, N=10_000, as_func=True) #c=0
w_dark, x_dark = soliton(c_2, k, soliton_type="dark", L=L, N=10_000, as_func=True)
```

## Black tests

### Varying $\Delta t$

```python
tf = 5
N = 200

w_1 = Real_CNFD(L, N, 0.1, tf, pars, "neumann")
w_1.set_init_condition(w_black)
w_1.solve()

w_2 = Real_CNFD(L, N, 0.01, tf, pars, "neumann")
w_2.set_init_condition(w_black)
w_2.solve()

w_3 = Real_CNFD(L, N, 0.001, tf, pars, "neumann")
w_3.set_init_condition(w_black)
w_3.solve()
```

```python
# Norm
w_1.plot(tf, label=r"$dt=0.1$")
w_2.plot(tf, label=r"$dt=0.01$")
w_3.plot(tf, label=r"$dt=0.001$")
plt.plot(x_black, norm(w_black(x_black)))
plt.legend()
plt.xlabel(r"$x$")
plt.ylabel(r"$|\psi|^2$")

plt.show()

# Imaginary

w_1.plot_imag(tf, label=r"$dt=0.1$")
w_2.plot_imag(tf, label=r"$dt=0.01$")
w_3.plot_imag(tf, label=r"$dt=0.001$")
plt.plot(x_black, np.imag(w_black(x_black)))

plt.xlabel(r"$x$")
plt.ylabel(r"Im$(\psi)$")
plt.legend()
plt.show()


# Real

w_1.plot_real(tf, label=r"$dt=0.1$")
w_2.plot_real(tf, label=r"$dt=0.01$")
w_3.plot_real(tf, label=r"$dt=0.001$")
plt.plot(x_black, np.real(w_black(x_black)))

plt.xlabel(r"$x$")
plt.ylabel(r"Re$(\psi)$")
plt.legend()
plt.show()
```

### Varying $N$

```python
dt = 0.001
tf = 3

w_1 = Real_CNFD(L, 500, dt, tf, pars, "neumann")
w_2 = Real_CNFD(L, 1000, dt, tf, pars, "neumann")
w_3 = Real_CNFD(L, 2000, dt, tf, pars, "neumann")

w_1.set_init_condition(w_black)
w_2.set_init_condition(w_black)
w_3.set_init_condition(w_black)

w_1.solve()
w_2.solve()
w_3.solve()
```

## Real and complex time comparison

### Black soliton: Varying $\Delta t$

```python
dt = 0.1
tf = 3
N = 500

w_comp = Complex_CNFD(L, N, dt, tf, pars, "neumann")
w_comp.set_init_condition(w_black)
w_comp.solve()

w_comp.plot(tf, label="Complex num")
plt.plot(x_black, norm(w_black(x_black)), label="Theo")
plt.legend()
plt.show()
w_comp.plot_imag(tf, label="Complex num")
plt.plot(x_black, np.imag(w_black(x_black)), label="Theo")
plt.legend()
plt.show()
w_comp.plot_real(tf, label="Complex num")
plt.plot(x_black, np.real(w_black(x_black)), label="Theo")
plt.legend()
plt.show()
```

```python
dts = np.array([2**(-j) for j in range(1, 8)])
N = 500

total = 10

means_real = []
means_comp = []

for dt in dts:
    times_real = np.zeros(total)
    times_comp = np.zeros(total)

    for j in range(total):
        w_real = Real_CNFD(L, N, dt, tf, pars, "neumann")
        w_real.set_init_condition(w_black)
        w_real.solve()
        t_r = w_real.get_execution_time()
        times_real[j] = t_r

    for j in range(total):
        w_comp = Complex_CNFD(L, N, dt, tf, pars, "neumann")
        w_comp.set_init_condition(w_black)
        w_comp.solve()
        t_c = w_comp.get_execution_time()
        times_comp[j] = t_c

    means_real += [times_real.mean()]
    means_comp += [times_comp.mean()]
```

```python
plt.plot(dts, means_real, ".-", label="Real NR")
plt.plot(dts, means_comp, ".-", label="Complex NR")
plt.yscale("log")
plt.xscale("log")

plt.xlabel("$\log(\Delta t)$")
plt.ylabel("$\log(T_{ex})$")
plt.legend()
#plt.savefig("Img/Times_comparison/dt_static.png", dpi=300, bbox_inches = "tight")
plt.show()
```
